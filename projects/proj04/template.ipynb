## Step 2: Data Cleaning and Exploratory Data Analysis

```python
# TODO
import pandas as pd
import matplotlib.pyplot as plt

teams_df = pd.read_csv("2022_LoL_esports_match_data_from_OraclesElixir.csv")

obj_cols   = ['firstblood', 'firstdragon', 'firstherald', 'firstbaron', 'firsttower']
obj_labels = ['First Blood', 'First Dragon', 'First Herald', 'First Baron', 'First Tower']

win_rates = [
    teams_df[teams_df[col] == True]['result'].mean() * 100
    for col in obj_cols
]

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(obj_labels, win_rates, color='#0BC4C4', edgecolor='white', width=0.55)
ax.axhline(50, color='gray', linestyle='--', linewidth=1, label='50% baseline')
ax.set_title("Win Rate When Securing First Objective", fontweight='bold')
ax.set_ylabel("Win Rate (%)")
ax.set_ylim(40, 95)
ax.legend()

for bar, val in zip(bars, win_rates):
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.8,
            f"{val:.1f}%", ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```
